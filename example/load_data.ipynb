```python
import urllib.request
from tqdm import tqdm
from pathlib import Path

data_url_list = [
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/target_series.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/fin/target_series_extended.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/fin/calendar_extended.csv',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/fin/sample_submit_extended.csv',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_1.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_2.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_3.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_4.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/fin/transactions_5.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_extra_1.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_extra_2.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_extra_3.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/transactions_extra_4.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/fin/transactions_extra_5.parquet',
    'https://storage.yandexcloud.net/vtb-competition-25-private/Task2/fin/profiles_extended.parquet',
]

DATAPATHH = Path('../data')
DATAPATHH.mkdir(parents=True, exist_ok=True)
```

```python
def show_progress(block_num, block_size, total_size):
    print(round(block_num * block_size / total_size *100,2), end="\r")

for url in tqdm(data_url_list):
    filepath = DATAPATHH / Path(url).name
    if not filepath.exists():
        try:
            urllib.request.urlretrieve(url, filepath, show_progress)
        except:
            print(f'error: {url}')
```
